1. Positivo, negativo ou zero

```python
while True:
  try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
    numero = int(input("Digite um número: "))
    break
  except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
    print("Digite um número inteiro válido.")
if numero > 0:
  print("Positivo")
elif numero < 0:
  print("Negativo")
else:
  print("Zero")
```

2. Maior entre dois números

```python
valores = []
for i in range(2):
  while True:
    try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
      n = int(input(f"Digite o {i+1}º número: "))
      valores.append(n)
      break
    except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
      print("Informe um inteiro válido")
a, b = valores
if a > b:
  print("Maior:", a)
elif b > a:
  print("Maior:", b)
else:
  print("Os números são iguais.")
```

5. Situação do aluno

```python
notas = []
for i in range(2):
  while True:
    try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
      nota = float(input(f"Digite a {i+1}ª nota: "))
      if 0 <= nota <= 10:
        notas.append(nota)
        break
      print("A nota deve estar entre 0 e 10.")
    except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
      print("Digite uma nota numérica válida.")
media = sum(notas) / 2
if media >= 7:
  print("Aprovado")
elif media >= 5:
  print("Recuperação")
else:
  print("Reprovado")
print("Média:", media)
```

9. Contando números pares

Leia exatamente 10 inteiros e conte quantos são pares. Se a entrada for inválida, solicite novamente a mesma posição.

```python
numeros = []
for i in range(10):
  while True:
    try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
      N = int(input(f"Digite o {i + 1}° inteiro: "))
      numeros.append(N)
      break
    except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
      print("Digite um número inteiro.")
pares = 0
for numero in numeros:
  if numero % 2 == 0:
    pares += 1
print(f"A quantidade de pares é: {pares}")
```

13. Tentativas de senha

Senha correta: 1234. Continue solicitando até acertar. Trate entradas não inteiras caso escolha comparar a senha como inteiro.

```python
senha_correta = 1234
senha_certa = False
while not senha_certa:
  try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
    senha = int(input("Digite a senha: "))
    if senha == senha_correta:
      senha_certa = True
      break
    print("Senha incorreta. Tente novamente.")
  except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
    print("Digite uma tentativa válida.")
print("Parabéns! Acesso concedido.")
```

14. Número secreto

O número secreto é 7. Dê dicas (maior/menor), conte tentativas e trate entradas não inteiras. Defina se entrada inválida conta como tentativa.

```python
num_secreto = 7
tentativas = 0
num_certo = False
while not num_certo:
  try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
    tentativas += 1 #entradas inválidas contam como tentativa
    num = int(input("Digite um número: "))
    if num == num_secreto:
      break
    elif num_secreto < num:
      print("O número secreto é menor. Tente novamente.")
    else:
      print("O número secreto é maior. Tente novamente.")
  except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
    print("Digite um número inteiro.")
print("Parabéns! Você acertou o número secreto (7)!")
print(f"Tentativas: {tentativas}")
```

15. Analisando uma sequência

Leia inteiros até 0; informe quantidade, positivos, negativos e soma. Entrada inválida não encerra nem altera os resultados.

```python
numeros = []
while True:
  try: #o código é estruturado para caso haver uma exceção, ela ocorrer antes do break, assim o usuário pode repetir
    N = int(input("Digite um número inteiro (digite 0 para parar): "))
    if N != 0:
      numeros.append(N)
    else:
      break
  except ValueError: #caso o usuário digite uma string ou um float inválido, como 3.14 ou "abc", esse except trata da exceção ValueError que ocorreria
    print("Digite um número válido.")
quantidade = 0
positivos = 0
negativos = 0
soma = 0
for numero in numeros:
  quantidade += 1
  soma += numero
  if numero > 0:
    positivos += 1
  else:
    negativos += 1
print(f"Quantidade de números: {quantidade}")
print(f"Quantidade de números positivos: {positivos}")
print(f"Quantidade de números negativos: {negativos}")
print(f"Soma dos números: {soma}")
```
