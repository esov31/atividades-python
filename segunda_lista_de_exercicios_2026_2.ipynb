4. Estacionamento

Um estacionamento cobra 8,00 reais por até 2 horas. Acima de 2 e até 5 horas, cobra 8,00 reais mais 3,00 reais por hora adicional. Acima de 5 horas, cobra 17,00 reais mais 2,00 reais por hora que ultrapassar 5. Leia a quantidade inteira de horas e calcule o valor a pagar.

```python
horas = int(input("Digite a quantidade total de horas: "))
if horas <= 0:
  print("Quantidade de horas inválida.")
elif horas <= 2:
  valor = 8
elif 2 < horas <=5:
  valor = 8 + 3 * (horas - 2)
elif horas > 5:
  valor = 17 + 2 * (horas - 5)
print(f"O valor total a ser pago é: {valor} reais.")
```

5. Máquina de venda de ingresso

Um ingresso custa 25,00 reais. Solicite valores inteiros ao usuário até que o total inserido seja suficiente para a compra. Ao final, informe o total inserido e o troco.

```python
valor = int(input("Digite o valor do pagamento: "))
while valor < 25:
  valor = int(input("Valor insuficiente. Digite o valor do pagamento: "))

print(f"Total inserido: {valor} reais")
print(f"Troco: {valor - 25} reais")
```

6. Maior valor e sua posição

vez.

8. Detectando repetição consecutiva

Leia 8 números inteiros e informe se em algum momento foram digitados dois valores iguais consecutivamente. Não é necessário informar quais valores foram iguais.

```python
N = int(input("Digite um número inteiro: "))
t = N
valores_consecutivos = False
for i in range(7):
  N = int(input("Digite um número inteiro: "))
  if N == t:
    valores_consecutivos = True
  t = N
if valores_consecutivos:
  print("Foram digitados dois valores iguais consecutivamente.")
else:
  print("Não foram digitados dois valores iguais consecutivamente.")
```

9. Controle de saldo

O programa começa com saldo de R$ 500,00. Serão realizadas 6 operações. Em cada uma, leia o tipo (1 = depósito; 2 = saque) e o valor. Depósitos aumentam o saldo. Saques só podem ocorrer se houver saldo suficiente; caso contrário, mostre 'Saldo insuficiente'. Ao final, mostre o saldo.


```python
saldo = 500
for i in range(6):
  operacao = int(input("Digite o tipo de operação (1 = depósito; 2 = saque): "))
  if operacao == 1:
    deposito = int(input("Digite o valor a ser depositado: "))
    saldo += deposito
  elif operacao == 2:
    saque = int(input("Digite o valor a ser sacado: "))
    if saldo - saque >= 0:
      saldo -= saque
    else:
      print("Saldo insuficiente.")
  else:
    print("Operação inválida.")
print(f"O saldo resultante é: {saldo} reais.")
```

10. Caixa eletrônico simplificado.

Leia um valor inteiro positivo e informe a menor quantidade de cédulas para representá-lo, usando R$ 100, 50, 20, 10, 5, 2 e 1.

```python
valor = int(input("Digite um valor inteiro positivo: "))
cedulas = [100, 50, 20, 10, 5, 2, 1]
quantidade_de_cedulas = 0
for cedula in cedulas:
    quantidade_de_cedulas += valor // cedula
    valor %= cedula
print(f"A menor quantidade de cédulas é: {quantidade_de_cedulas}")
```

11. Número primo

Leia um inteiro maior que 1 e informe se ele é primo.

```python
N = int(input("Digite um número inteiro maior do que 1: "))
raiz = int(N ** 0.5)
primo = True
for i in range(2, raiz + 1): #Não é necessário verificar se o número é divisível por cada um de seus antecessores, basta considerar até a raíz quadrada desse número (ou seu valor aproximado, se não for um quadrado perfeito), visto que um número é composto de uma dupla de fatores, sendo um menor que sua raíz e outro maior, de tal modo que se não há divisor antes da raís do número, não haverá depois também.
  if N % i == 0:
    primo = False
    break
if primo:
  print("O número é primo.")
else:
  print("O número não é primo.")
```

12. Divisores de um número

Leia um inteiro positivo N, mostre todos os seus divisores positivos e informe quantos divisores ele possui.

```python
N = int(input("Digite um número inteiro positivo: "))
divisores = []
n_divisores = 0
for i in range(1, N + 1):
  if N % i == 0:
    divisores.append(i)
    n_divisores += 1
print(f"Divisores de {N}: {divisores}. \nNúmero de divisores de {N}: {n_divisores}.")
```

13. Números primos em um intervalo

Leia A e B positivos, com A < B, mostre os primos entre A e B (inclusive) e informe quantos foram encontrados.

```python
A = int(input("Digite um número inteiro positivo: "))
B = int(input("Digite um número inteiro positivo maior que o anterior: "))
primos = []
n_primos = 0
for numero in range(A, B + 1):
  primo = True
  if numero < 2:
    primo = False
  else:
    raiz_n = int(numero ** 0.5)
    for i in range(2, raiz_n + 1):
      if numero % i == 0:
        primo = False
        break
    if primo:
      primos.append(numero)
      n_primos += 1
print(f"Primos entre {A} e {B}: {primos}. \nNúmero de primos entre {A} e {B}: {n_primos}.")
```

14. Jogo com limites de tentativas

Considere o número secreto 37. O usuário terá no máximo 5 tentativas. Após cada erro, informe se o secreto é maior ou menor que o palpite. Termine ao acertar ou ao usar as 5 tentativas.

```python
n_secreto = 37
for i in range(5):
  n_palpite = int(input("Digite o palpite: "))
  if n_palpite == n_secreto:
    print("Parabéns! Você venceu!")
    break
  if i < 4:
    if n_palpite < n_secreto:
      print("Tente novamente, o número secreto é maior.")
    else:
      print("Tente novamente, o número secreto é menor.")
  if i == 4:
    print("Você perdeu!")
```

15. Desafio final

Um caixa inicia o dia com R$ 200,00. Leia operações até que o código 0 seja informado. Código 1 = entrada de dinheiro; código 2 = retirada. Para cada operação válida, leia o valor. Uma retirada só é permitida se houver saldo suficiente. Ao final, informe: saldo final, quantidade de entradas, quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total que saiu.

```python
saldo = 200
q_entradas = 0
q_retiradas_realizadas = 0
q_retiradas_recusadas = 0
entradas_t = 0
retiradas_t = 0
codigo = 1
while codigo != 0:
  codigo = int(input("Digite o número da operação. Código 0 = Encerrar; Código 1 = Entrada de dinheiro; Código 2 = Retirada de dinheiro: "))
  if codigo == 1:
    entrada = round(float(input("Digite o valor da entrada: ")), 2)
    if entrada <= 0:
      print("Entrada inválida. Digite um valor maior que zero. ")
    else:
      q_entradas += 1
      entradas_t += entrada
      saldo += entrada
  elif codigo == 2:
    retirada = round(float(input("Digite o valor da retirada: ")), 2)
    if retirada > saldo:
      q_retiradas_recusadas += 1
      print("Retirada inválida. O saldo é menor que o valor requisitado.")
    else:
      q_retiradas_realizadas += 1
      retiradas_t += retirada
      saldo -= retirada
print(f"Saldo final: {round(saldo, 2)}. \nQuantidade de entradas: {q_entradas}. \nQuantidade de retiradas realizadas: {q_retiradas_realizadas}. \nQuantidade de retiradas reacusadas: {q_retiradas_recusadas}. \nTotal que entrou: {round(entradas_t, 2)}. \nTotal que saiu: {round(retiradas_t, 2)}.")
```
