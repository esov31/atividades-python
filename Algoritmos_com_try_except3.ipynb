1. Leia 5 notas, armazene todas em uma lista e, somente depois das leituras, mostre as notas, a média e quantas notas são maiores ou iguais à média.

```python
notas = []
for i in range(5):
  while True:
    try:
      nota = float(input(f"Digite a {i+1}ª nota: "))
      if not 0 <= nota <= 10:
        print("A nota deve estar entre 0 e 10.")
        continue
      notas.append(nota)
      break
    except ValueError:
      print("Entrada inválida. Digite um número.")
soma = 0
for nota in notas:
  soma += nota
media = soma / len(notas)
acima_ou_igual = 0
for nota in notas:
  if nota >= media:
    acima_ou_igual += 1
print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

3. Leia nome, preço e quantidade de um produto. Armazene os dados em um dicionário e mostre o cadastro e o valor total em estoque.

```python
produto = {"nome": 0, "preço": 0, "quantidade": 0}
produto["nome"] = input("Digite o nome do produto: ")
while True:
  try:
    produto["preço"] = float(input("Digite o preço do produto: "))
    produto["quantidade"] = int(input("Digite a quantidade do produto: "))
    if produto['preço'] < 0 or produto['quantidade'] < 0:
      print("Digitye valores positivos.")
      continue
    break
  except ValueError:
    print("Digite um preço válido.")
valor_em_estoque = produto["preço"] * produto["quantidade"]
print(f"O valor total em estoque é: {valor_em_estoque}")
```

4. Leia 10 números inteiros e armazene todos em uma lista. Depois das 10 leituras, percorra a lista e crie duas novas listas: uma apenas com os pares e outra apenas com os ímpares. Mostre as três listas.

```python
lista_de_numeros = []
for i in range(10):
  while True:
    try:
      num = int(input(f"Digite o {i + 1}° número: "))
      lista_de_numeros.append(num)
      break
    except ValueError:
      print('Digite um valor numérico válido.')
pares = []
impares = []
for num in lista_de_numeros:
  if num % 2 == 0:
    pares.append(num)
  else:
    impares.append(num)
print(f"Números: {lista_de_numeros}")
print(f"Pares: {pares}")
print(f"Ímpares: {impares}")
```

6. Leia nomes até que o usuário digite FIM. Armazene somente nomes que ainda não estejam na lista. Ao final, mostre a lista e a quantidade de nomes diferentes cadastrados.

```python
FIM = False
nomes = []
while not FIM:
  nome = input("Digite o nome (FIM para encerrar): ")
  if nome != 'FIM':
    if nome not in nomes:
      nomes.append(nome)
    else:
      print("Digite um nome que ainda não está na lista.")
  else:
    FIM = True
print(f"Lista de nomes: {nomes}")
print(f"Quantidade de nomes: {len(nomes)}")
```

7. Crie uma tupla contendo os sete dias da semana, em ordem. Leia um número de 1 a 7 e mostre o dia correspondente. Caso o número esteja fora desse intervalo, informe que a opção é inválida

```python
semana = ("domingo", "segunda-feira", "terça-feira", "quarta-feira", "quinta-feira", "sexta-feira", "sábado")
while True:
  try:
    dia = int(input("Digite o número da semana (de 1 a 7): "))
  except ValueError:
    print("Digite um valor válido.")
    continue
  if dia < 1 or dia > 7:
    print("Digite um número no intervalo.")
    continue
  break
print(f"O dia correpondente é {semana[dia - 1]}")
```

9. Cadastre nome, turma, nota1 e nota2 de um aluno em um dicionário. Calcule a média, acrescente ao próprio dicionário as chaves media e situacao e mostre todos os dados. Use: aprovado >= 7; recuperação >= 5; reprovado < 5.

```python
dicionario = {"nome": 0, "turma": 0, "nota 1": 0, "nota 2": 0}
dicionario["nome"] = input("Digite o nome do aluno: ")
dicionario["turma"] = input("Digite a turma do aluno: ")
while True:
  try:
    nota1 = float(input("Digite o valor da primeira nota: "))
    nota2 = float(input("Digite o valor da segunda nota: "))
    if nota1 > 10 or nota1 < 0 or nota2 > 10 or nota2 < 0:
      print("Digite uma nota válida.")
      continue
    break
  except ValueError:
    print("Digite um valor válido.")
dicionario["nota 1"] = nota1
dicionario["nota 2"] = nota2
media = (nota1 + nota2) / 2
if media >= 7:
  situacao = "aprovado"
elif media >= 5:
  situacao = "recuperação"
else:
  situacao = "reprovado"
dicionario["média"] = media
dicionario["situação"] = situacao
print(dicionario)
```

12. Crie um dicionário com 4 produtos e suas quantidades em estoque. Leia o nome de um produto e uma quantidade vendida. Se o produto existir e houver estoque suficiente, atualize o dicionário; caso contrário, informe o problema. Mostre o estoque final.

```python
produtos = {"Maçã": 50, "Banana": 60, "Uva": 120, "Melancia": 100}
nome_produto = input("Digite o nome do produto: ")
while True:
  try:
    quantidade_vendida = int(input("Digite a quantidade vendida: "))
    if quantidade_vendida < 0:
      print("Digite uma quantidade válida.")
      continue
    break
  except ValueError:
    print("Digite uma quantidade válida.")
if nome_produto not in produtos:
  print("Produto não encontrado.")
elif produtos[nome_produto] < quantidade_vendida:
  print("Quantidade insuficiente.")
else:
  produtos[nome_produto] = produtos[nome_produto] - quantidade_vendida
print(produtos)
```

13. Cadastre 5 alunos. Para cada aluno, leia nome e duas notas, calcule a média e armazene nome, notas e média em um dicionário. Cada dicionário deve ser adicionado a uma lista. Ao final, mostre nome e média de todos os alunos.

```python
alunos = []
for i in range(5):
  nome = input("Digite o nome do aluno: ")
  while True:
    try:
      nota1 = int(input("Digite a primeira nota do aluno: "))
      nota2 = int(input("Digite a segunda nota do aluno: "))
      if nota1 < 0 or nota1 > 10 or nota2 < 0 or nota2 > 10:
        print("Digite uma nota válida.")
      break
    except ValueError:
      print("Digite um valor válido.")
  media = (nota1 + nota2) / 2
  aluno = {"nome": nome, "nota 1": nota1, "nota 2": nota2, "média": media}
  alunos.append(aluno)
for aluno in alunos:
  print(f"{aluno['nome']}: {aluno['média']}")
```

14. Cadastre 5 produtos usando tuplas no formato (codigo, nome, preco) e armazene essas tuplas em uma lista. Depois, leia um código e procure o produto correspondente, mostrando nome e preço. Informe quando o código não for encontrado.

```python
produtos = []
for i in range(5):
  codigo = input("Digite o código do produto: ")
  nome = input("Digite o nome do produto: ")
  while True:
    try:
      preco = int(input("Digite o preço do produto: "))
      break
    except ValueError:
      print("Digite um valor válido.")
  produto = (codigo, nome, preco)
  produtos.append(produto)
codigo_procurado = input("Digite o código a ser procurado: ")
codigo_existe = False
for produto in produtos:
  if codigo_procurado in produto:
    codigo_existe = True
    break
if codigo_existe:
  for i in range(len(produtos)):
    if produtos[i][0] == codigo_procurado:
      print(f"Nome: {produtos[i][1]}; Preço: {produtos[i][2]}")
      break
else:
  print("Código não encontrado.")
```

15. Construa um pequeno cadastro de 5 itens de estoque. Cada item deve ser um dicionário com codigo, nome, quantidade e preco. Armazene os dicionários em uma lista. Depois: (a) mostre todos os itens; (b) calcule o valor total do estoque; (c) identifique o item de maior valor em estoque (quantidade ×preço); (d) crie uma tupla resumo contendo(quantidade_de_itens_cadastrados, valor_total_estoque).

```python
produtos = []
for i in range(5):
  codigo = input("Digite o código do produto: ")
  nome = input("Digite o nome do produto: ")
  while True:
    try:
      quantidade = int(input("Digite a quantidade do produto: "))
      preco = int(input("Digite o preço do produto: "))
      break
    except ValueError:
      print("Digite um valor válido.")
  produto = {"codigo": codigo, "nome": nome, "quantidade": quantidade, "preço": preco}
  produtos.append(produto)
print(produtos)
valor_total = 0
for produto in produtos:
  valor_total += produto["quantidade"] * produto["preço"]
print(f"Valor total em estoque: {valor_total}")
maior_valor = produtos[0]["quantidade"] * produtos[0]["preço"]
for produto in produtos:
  if (produto["quantidade"] * produto["preço"]) > maior_valor:
    item_de_maior_valor = produto
print(f"Item de maior valor: {item_de_maior_valor}")
quantidade_de_itens_cadastrados = 0
for produto in produtos:
  quantidade_de_itens_cadastrados += produto["quantidade"]
resumo = (quantidade_de_itens_cadastrados, valor_total)
print(resumo)
```
