Leia 7 temperaturas, armazene todas em uma lista e, somente depois das leituras, mostre a média, quantas ficaram acima, abaixo e iguais à média.

```python
temperaturas = []
for i in range(7):
 temperaturas.append(float(input("Temperatura: ")))
soma = 0
for t in temperaturas:
 soma += t
media = soma / len(temperaturas)
acima = abaixo = iguais = 0
for t in temperaturas:
 if t > media:
  acima += 1
 elif t < media:
  abaixo += 1
 else:
  iguais += 1
print("Temperaturas:", temperaturas)
print("Média:", media)
print("Acima:", acima, "Abaixo:", abaixo, "Iguais:", iguais)
```

```python
lista_n = []
q_e_c_maior = 0
q_e_c = 1
for i in range(10):
  numero = int(input("Digite um número inteiro: "))
  lista_n.append(numero)
for i in range(9):
  if lista_n[i + 1] > lista_n[i]:
    q_e_c += 1
    if q_e_c > q_e_c_maior:
      q_e_c_maior = q_e_c
  elif lista_n[i] > lista_n[i - 1] + 1:
    q_e_c = 1
print(f"O maior número de elementos consecultivos em ordem estritamento crescente é: {q_e_c_maior}.")
```

5. Pontos no plano

Cadastre 6 pontos como tuplas (x, y) em uma lista. Informe quantos estão em cada quadrante e qual ponto possui maior valor de x² + y².

```python
lista_pontos = []
q_1_quadrante = 0
q_2_quadrante = 0
q_3_quadrante = 0
q_4_quadrante = 0
q_origem = 0
q_eixo_x = 0
q_eixo_y = 0
for i in range(6):
  x = int(input("Digite o valor do ponto no eixo das abcissas: "))
  y = int(input("Digite o valor do ponto no eixo das ordenadas: "))
  print("----------")
  ponto = (x, y)
  lista_pontos.append(ponto)
mais_distante = lista_pontos[0]
for ponto in lista_pontos:
  if ponto[0] == 0 and ponto[1] == 0:
    q_origem += 1
  elif ponto[0] == 0:
    q_eixo_y += 1
  elif ponto[1] == 0:
    q_eixo_x += 1
  elif ponto[0] > 0 and ponto[1] > 0:
    q_1_quadrante += 1
  elif ponto[0] < 0 and ponto[1] > 0:
    q_2_quadrante += 1
  elif ponto[0] < 0 and ponto[1] < 0:
    q_3_quadrante += 1
  else:
    q_4_quadrante += 1
for ponto in lista_pontos:
  if ((ponto[0] ** 2) + (ponto[1] ** 2)) > ((mais_distante[0] ** 2) + (mais_distante[1] ** 2)):
    mais_distante = ponto
print(f"Quantidade de pontos na origem: {q_origem}.")
print(f"Quantidade de pontos no eixo y: {q_eixo_y}.")
print(f"Quantidade de pontos no eixo x: {q_eixo_x}.")
print(f"Quantidade de pontos no primeiro quandrante: {q_1_quadrante}.")
print(f"Quantidade de pontos no segundo quadrante: {q_2_quadrante}.")
print(f"Quantidade de pontos no terceiro quadrante: {q_3_quadrante}.")
print(f"quantidade de pontos no quarto quadrante: {q_4_quadrante}.")
print(f"O ponto mais distante da origem: {mais_distante}.")
```

6. Frequência e palavra mais repetida

Leia 10 palavras e construa um dicionário de frequências. Mostre as quantidades e todas as palavras empatadas na maior frequência.

```python
lista_palavras = []
palavras_empate =[]
for i in range(10):
  palavra = input("Digite uma palavra: ")
  lista_palavras.append(palavra)
dic_f = {}
for word in lista_palavras:
  f = lista_palavras.count(word)
  dic_f[word] = f
maior_f = 1
for word in lista_palavras:
  if maior_f < dic_f[word]:
    maior_f = dic_f[word]
for word in lista_palavras:
  if maior_f == dic_f[word] and word not in palavras_empate:
    palavras_empate.append(word)
print(f"A maior frequência é: {maior_f}. \nAs palavras empatadas na maior frequência são: {palavras_empate}.")
```

8. Controle de estoque com movimentações

Cadastre 5 produtos com codigo, nome, preco e quantidade. Depois realize 5 operações de entrada (E) ou saída (S). Ao final, mostre o estoque atualizado e seu valor financeiro total.

```python
produtos =[]
for i in range(5):
  codigo = int(input("Digite o número do código do produto: "))
  nome = input("Digite o nome do produto: ")
  preco = float(input("Digite o preço do produto: "))
  quantidade = int(input("Digite a quantidade do produto: "))
  print("----------")
  produto = {"codigo": codigo, "nome": nome, "preco": preco, "quantidade": quantidade}
  produtos.append(produto)
for i in range(5):
  codigo_in = False
  codigo_d = int(input("Digite o código do produto que buscas: "))
  for produto in produtos:
    if produto["codigo"] == codigo_d:
      produto_procurado = produto
      codigo_in = True
      break
  if codigo_in is False:
    print("Código inválido. Digite novamente.")
    continue

  operacao = input("Digite o tipo de operação (para Entrada, digite E; para Saída, digite S): ")
  if operacao == 'E':
    q_entrada = int(input("Digite a quantidade de entrada: "))
    if q_entrada < 0:
      print("Quantidade de entrada inválida.")
      print("-----------")
      continue
    else:
      produto_procurado["quantidade"] += q_entrada
  elif operacao == 'S':
    q_saida = int(input("Digite a quantidade de retirada: "))
    if q_saida > produto_procurado["quantidade"]:
      print("Quantidade em estoque insuficiente.")
      print("-----------")
      continue
    elif q_saida < 0:
      print("Quantidade de saída inválida.")
      print("-----------")
      continue
    else:
      produto_procurado["quantidade"] -= q_saida
  else:
    print("Operação inválida.")
    print("-----------")
    continue
  print("----------")
valor_t = 0
for produto in produtos:
  valor_t += (produto["quantidade"] * produto["preco"])
print("Estoque atualizado: ")
for produto in produtos:
  print(f"Nome do produto: {produto['nome']}. Quantidade em estoque: {produto['quantidade']}.")
print(f"Valor total em estoque: {valor_t}")
```

9. Ranking de jogadores

Cadastre 5 jogadores e suas pontuações em 3 partidas. Informe total, média, jogador(es) com maior total e jogador(es) mais regulares, considerando a menor diferença entre maior e menor pontuação.

```python
jogadores = []
for i in range(5):
  nome = input("Digite o nome do jogador: ")
  pont1 = int(input(f"Digite a primeira pontuação do jogador: "))
  pont2 = int(input(f"Digite a segunda pontuação do jogador: "))
  pont3 = int(input(f"Digite a terceira pontuação do jogador: "))
  print("----------")
  jogador = {"nome": nome, "pontuação 1": pont1, "pontuação 2": pont2, "pontuação 3": pont3}
  jogadores.append(jogador)
for jogador in jogadores:
  total = 0
  media = 0
  total = (jogador["pontuação 1"] + jogador["pontuação 2"] + jogador["pontuação 3"])
  media = round((total / 3), 2)
  jogador["total"] = total
  jogador["média"] = media
  print(f"Jogador: {jogador['nome']}. Total: {jogador['total']}. Média: {jogador['média']}.")
print("----------")
maior_t = jogadores[0]["total"]
for jogador in jogadores:
  if jogador["total"] >= maior_t:
    maior_t = jogador["total"]
print("Jogador(es) com maior total: ")
for jogador in jogadores:
  if jogador["total"] == maior_t:
    print(jogador["nome"])
print("----------")
diferencas = []
for jogador in jogadores:
  pont = [jogador["pontuação 1"], jogador["pontuação 2"], jogador["pontuação 3"]]
  maior_p = pont[0]
  menor_p = pont[0]
  for pontuação in pont:
    if pontuação > maior_p:
      maior_p = pontuação
    if pontuação < menor_p:
      menor_p = pontuação
  diferenca = maior_p - menor_p
  diferencas.append(diferenca)
menor_d = diferencas[0]
for diferenca in diferencas:
  if diferenca <= menor_d:
    menor_d = diferenca
print("Jogador(es) mais regulares: ")
for i in range(5):
  if diferencas[i] == menor_d:
    print(jogadores[i]["nome"])
```

10. Agenda de contatos sem duplicidade

Cadastre contatos até que seja digitado FIM. Cada contato deve possuir nome e telefone. Não permita dois contatos com o mesmo nome. Depois, permita consultar nomes e informe o telefone quando o contato existir.

```python
FIM = False
contatos = []
while FIM is False:
  nome_repetido = False
  nome = input("Digite o nome do contato: ")
  telefone = (input("Digite o telefone do contato: "))
  contato = {"nome": nome, "telefone": telefone}
  for contato_cadastrado in contatos:
    if contato_cadastrado["nome"] == nome:
      nome_repetido = True
      break
  if not nome_repetido:
    contatos.append(contato)
  confirmacao = input("Deseja continuar (S/N)?: ")
  print("----------")
  if confirmacao == "N":
    FIM = True
nome_consulta = input("Digite o nome do contato que deseja consultar: ")
nome_existe = False
for contato in contatos:
  if contato["nome"] == nome_consulta:
    print(f"Telefone do contato: {contato['telefone']}")
    nome_existe = True
    break
if not nome_existe:
  print("Contato não encontrado.")
```

11. Pesquisa de votos

Leia votos para 4 opções identificadas por códigos 1 a 4 até que seja digitado 0. Conte os votos em um dicionário e, ao final, mostre o total de cada opção, o percentual de cada uma e a(s) opção(ões) vencedora(s)

```python
votos = {"1": 0, "2": 0, "3": 0, "4": 0}
opcoes = ["1", "2", "3", "4"]
codigo = 5
while codigo != "0":
  codigo = input("Digite o código da opção: ")
  if codigo in opcoes:
    votos[codigo] += 1
  elif codigo not in opcoes and codigo != "0":
    print("Código inválido")
  print("----------")
total = votos["1"] + votos["2"] + votos["3"] + votos["4"]
for opcao in opcoes:
  print(f"Opção {opcao}: {votos[opcao]} votos. \nPercentual: {round((votos[opcao] / total) * 100, 2)}%.")
print("----------")
maior_opcao = votos["1"]
print("Opções vencedoras: ")
for opcao in opcoes:
  if votos[opcao] > maior_opcao:
    maior_opcao = votos[opcao]
for opcao in opcoes:
  if votos[opcao] == maior_opcao:
    print(opcao)
```

12. Histórico de movimentações

Crie um estoque inicial em dicionário e realize 8 movimentações. Cada movimentação deve ser uma tupla (codigo, tipo, quantidade) armazenada em uma lista. Mostre estoque final e histórico

```python
estoque = {1: 100, 2: 90, 3: 80, 4: 70, 5: 60}
movimentacoes = []
for i in range(8):
  cod_mov = int(input("Digite o código do produto: "))
  if cod_mov not in range(1, 6):
    print("Código inválido.")
    movimentacao = (cod_mov, "N/A", "N/A", "RECUSADA")
    movimentacoes.append(movimentacao)
    print("----------")
    continue
  else:
    tipo = input("Digite o tipo de operação ('E' para entrada; 'S' para saída): ")
    if tipo == 'S':
      quantidade = int(input("Digite a quantidade de saída: "))
      if quantidade <= estoque[cod_mov] and quantidade > 0:
        movimentacao = (cod_mov, tipo, quantidade)
        movimentacoes.append(movimentacao)
        estoque[cod_mov] -= quantidade
        print("----------")
      else:
        print("Quantidade inválida.")
        movimentacao = (cod_mov, tipo, quantidade, "RECUSADA")
        movimentacoes.append(movimentacao)
        print("----------")
        continue
    elif tipo == 'E':
      quantidade = int(input("Digite a quantidade de entrada: "))
      if quantidade > 0:
        movimentacao = (cod_mov, tipo, quantidade)
        movimentacoes.append(movimentacao)
        estoque[cod_mov] += quantidade
        print("----------")
      else:
        print("Quantidade inválida.")
        movimentacao = (cod_mov, tipo, quantidade, "RECUSADA")
        movimentacoes.append(movimentacao)
        print("----------")
        continue
    else:
      print("Tipo de movimentação inválido.")
      movimentacao = (cod_mov, tipo, "N/A", "RECUSADA")
      movimentacoes.append(movimentacao)
      print("----------")
      continue
print(f"Estoque final: {estoque}")
print(f"Histórico: {movimentacoes}")
```

15. Sistema de vendas e relatório final

Cadastre 5 produtos com código, nome, preço e quantidade. Processe vendas até código 0. Uma venda só ocorre se o produto existir e houver estoque. Registre cada venda em uma tupla. Ao final, mostre estoque restante, unidades vendidas, valor total, produto(s) de maior valor em vendas, produtos sem venda, histórico e uma tupla-resumo (quantidade_de_vendas, unidades_vendidas, valor_total).

```python
produtos = []
for i in range(5):
  codigo = input("Digite o código do produto: ")
  nome = input("Digite o nome do produto: ")
  preco = int(input("Digite o preço do produto: "))
  quantidade = int(input("Digite a quantidade em estoque: "))
  print("---------")
  produto = {
      "código": codigo,
      "nome": nome,
      "preço": preco,
      "quantidade": quantidade
  }
  produtos.append(produto)
vendas = []
quantidade_de_vendas = 0
unidades_vendidas = 0
valor_total = 0
codigo_venda = ""
while codigo_venda != "0":
  codigo_venda = input("Digite o código do produto vendido: ")
  if codigo_venda != "0":
    quantidade_vendida = int(input("Digite a quantidade vendida: "))
    venda_valida = False
    for i in range(5):
      if produtos[i]["código"] == codigo_venda and quantidade_vendida <= produtos[i]["quantidade"]:
        venda_valida = True
        vendas.append((codigo_venda, quantidade_vendida))
        produtos[i]["quantidade"] -= quantidade_vendida
        quantidade_de_vendas += 1
        unidades_vendidas += quantidade_vendida
        valor_total += quantidade_vendida * produtos[i]["preço"]
        break
    if not venda_valida:
        print("Venda inválida.")
  print("----------")
print("Estoque restante: ")
for produto in produtos:
  print(produto)
print("----------")
print(f"Unidades vendidas: {unidades_vendidas}")
print(f"Valor total: {valor_total}")
print("----------")
print("Produto(s) de maior valor em vendas: ")
if len(vendas) > 0:
  vendas_produtos = {}
  for venda in vendas:
    for produto in produtos:
      if produto["código"] == venda[0] and produto["código"] not in vendas_produtos:
        vendas_produtos[produto["código"]] = venda[1] * produto["preço"]
      elif produto["código"] == venda[0] and produto["código"] in vendas_produtos:
        vendas_produtos[produto["código"]] += venda[1] * produto["preço"]
  maior_valor_vendas = 0
  for codigo in vendas_produtos:
    if vendas_produtos[codigo] > maior_valor_vendas:
      maior_valor_vendas = vendas_produtos[codigo]
  for produto in produtos:
    if produto["código"] in vendas_produtos and vendas_produtos[produto["código"]] == maior_valor_vendas:
      print(produto["nome"])
else:
  print("Nenhuma venda foi realizada.")
print("----------")
print("Histórico: ")
for venda in vendas:
  print(vendas)
print("----------")
print("Resumo (quantidade de vendas, unidades vendidas, valor total): ")
print((quantidade_de_vendas, unidades_vendidas, valor_total))
```
