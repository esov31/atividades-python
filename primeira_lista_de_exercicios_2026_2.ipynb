5.  Leia duas notas de um aluno e calcule sua média. Informe: a) Aprovado, se a média for maior ou igual a 7; b) Recuperação, se a média for maior ou igual a 5 e menor que 7; c) Reprovado, se a média for menor que 5. Ao final, mostre também a média calculada.

```python
nota1 = float(input("Digite a primeira nota do aluno: "))
nota2 = float(input("Digite a segunda nota do aluno: "))
media = (nota1 + nota2) / 2
if nota1 > 10 or nota1 < 0:
  print("Nota 1 inválida.")
elif nota2 > 10 or nota2 < 0:
  print("Nota 2 inválida.")
else:
  print(f"Média: {media}")
  if media >= 7:
    print("Aprovado.")
  elif media >=5:
    print("Recuperação.")
  elif media < 5:
    print("Reprovado.")
```

8. Leia um número inteiro positivo N e calcule a soma de todos os números de 1 até N.

```python
N = int(input("Digite um inteiro positivo: "))
t = 0
for i in range(1, N + 1):
  t += i
print(t)
```

14. Considere que o número secreto é 7. Peça ao usuário que tente descobrir o número. Enquanto ele não acertar: a) se o número digitado for menor que 7, mostre "Tente um número maior"; b) se for maior que 7, mostre "Tente um número menor". Quando acertar, mostre: Parabéns! Você acertou. Além disso, informe quantas tentativas foram necessárias.

```python
num_secr = 7
tent = 0
n9 = int(input("Tente descobrir o número secreto: "))
while n9 != num_secr:
  tent += 1
  if n9 < 7:
    n9 = int(input("Tenter um número maior: "))
  else:
    n9 = int((input("Tente um número menor: ")))
print("Parabéns!")
print(f"Foram necessárias {tent} tentativas.")
```

5. Faça um programa que leia números inteiros repetidamente. O programa deverá parar quando o usuário digitar 0. O zero serve apenas para encerrar e não deve entrar nos cálculos. Ao final, informe: a) quantidade de números digitados; b) quantidade de números positivos; c) quantidade de números negativos; d)soma de todos os números digitados.

```python
quant = 0
num_pos = 0
soma = 0
while True:
  n10 = int(input("Digite um número inteiro: "))
  if n10 == 0:
    break
  quant += 1
  soma += n10
  if n10 > 0:
      num_pos += 1
print("Quantidade:", quant)
print("Positivos:", num_pos)
print("Negativos:", quant - num_pos)
print("Soma:", soma)
```
