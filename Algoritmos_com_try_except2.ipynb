1. Maior e menor entre vários valores

```python
def ler_inteiro():
  while True:
    try:
      return int(input("Digite um número: "))
    except ValueError:
      print("Entrada inválida. Digite um inteiro.")
primeiro = ler_inteiro()
maior = menor = primeiro
for i in range(5):
  numero = ler_inteiro()
  if numero > maior:
    maior = numero
  if numero < menor:
    menor = numero
print("Maior:", maior)
print("Menor:", menor)
```

2. Estatísticas de uma turma

```python

soma = 0
aprovados = 0
abaixo = 0
for i in range(5):
  while True:
    try:
      nota = float(input(f"Nota do aluno {i+1}: "))
      if not 0 <= nota <= 10:
        print("Nota deve estar entre 0 e 10.")
        continue
      break
    except ValueError:
      print("Digite uma nota numérica válida.")
  soma += nota
  if nota >= 7:
    aprovados += 1
  else:
    abaixo += 1
print("Média:", soma / 5)
print("Notas >= 7:", aprovados)
print("Notas < 7:", abaixo)
```

4. Um estacionamento cobra 8,00 por até 2 horas. Acima de 2 e até 5 horas, cobra 8,00 mais 3,00 por hora adicional. Acima de 5 horas, cobra 17,00 mais 2,00 por hora que ultrapassar 5. Leia a quantidade inteira de horas e calcule o valor a pagar.

```python
while True:
  try:
    horas = int(input("Digite a quantidade de horas: "))
    if horas <= 0: #se for menor ou igual a 0
      print("Digite uma quantidade de horas válida: maior do que 0.")
      continue
    elif horas <= 2: #se for maior qu zero e menor ou igual a 2
      valor = 8
    elif horas <= 5: #se for maior que 2 e menor ou igual a 5
      valor = 8 + 3 * (horas - 2)
    else: #se for maior que 5
      valor = 17 + 2 * (horas - 5)
    break
  except ValueError:
    print("Digite uma quantidade horas válidas: um número inteiro.")
print(f"Quantidade de horas: {horas} horas.")
print(f"Valor a pagar: {valor} reais.")
```

5. Um ingresso custa R$ 25,00. Solicite valores inteiros ao usuário até que o total inserido seja suficiente para a compra. Ao final, informe o total inserido e o troco.

```python
valor_total = 0
while True:
  try:
    valor = int(input("Digite o valor para pagamento: "))
    if valor <= 0:
      print("Digite um valor maior do que zero")
    elif valor > 0:
      valor_total += valor
    if valor_total >= 25:
      troco = valor_total - 25
      break
  except ValueError:
    print("Digite um número válido.")
print(f"Valor total: {valor_total} reais.")
print(f"Troco: {troco} reais.")
```

9. O programa começa com saldo de R$ 500,00. Serão realizadas 6 operações. Em cada uma, leia o tipo (1 = depósito; 2 = saque) e o valor. Depósitos aumentam o saldo. Saques só podem ocorrer se houver saldo suficiente; caso contrário, mostre 'Saldo insuficiente'. Ao final, mostre o saldo.

```python
saldo = 500
for i in range(6):
  print(f"---- OPERAÇÃO {i+1} ----")
  while True:
    try:
      operacao = int(input("Digite '1' para depósito e '2' para saque: "))
      if operacao == 1:
        while True:
          try:
            deposito = int(input("Digite o valor do depósito: "))
            saldo += deposito
            break
          except ValueError:
            print("Digite um valor válido.")
      elif operacao == 2:
        while True:
          try:
            saque = int(input("Digite o valor do saque: "))
            if saldo >= saque:
              saldo -= saque
            else:
              print("Saldo insuficiente.")
            break
          except ValueError:
            print("Digite um valor válido.")
      else:
        print("Digite uma operação válida.")
      break
    except ValueError:
      print("Digite um número válido.")
print(f"O saldo é: {saldo}")
```

10. Leia um valor inteiro positivo e informe a menor quantidade de cédulas para representá-lo, usando R$ 100, 50, 20, 10, 5, 2 e 1

```python
valor = -1
while valor < 0:
  try:
    valor = int(input("Digite um valor inteiro positivo: "))
    if valor < 0:
      print("Digite um valor válido.")
  except ValueError:
    print("Digite um valor válido.")
quantidade_de_cedulas = {"100": 0, "50": 0, "20": 0, "10": 0, "5": 0, "2": 0, "1": 0}
quantidade_de_cedulas["100"] = int((valor - (valor % 100)) / 100)
quantidade_de_cedulas["50"] = int(((valor % 100) - ((valor % 100) % 50)) / 50)
quantidade_de_cedulas["20"] = int((((valor % 100) % 50) - (((valor % 100) % 50) % 20)) / 20)
quantidade_de_cedulas["10"] = int(((((valor % 100) % 50) % 20) - ((((valor % 100) % 50) % 20) % 10)) / 10)
quantidade_de_cedulas["5"] = int((((((valor % 100) % 50) % 20) % 10) - (((((valor % 100) % 50) % 20) % 10) % 5)) / 5)
quantidade_de_cedulas["2"] = int(((((((valor % 100) % 50) % 20) % 10) % 5) - ((((((valor % 100) % 50) % 20) % 10) % 5) % 2)) / 2)
quantidade_de_cedulas["1"] = int(((((((valor % 100) % 50) % 20) % 10) % 5) % 2))
menor_quantidade_de_cedulas = 0
for i in quantidade_de_cedulas:
  menor_quantidade_de_cedulas += quantidade_de_cedulas[i]
for i in quantidade_de_cedulas:
  print(f"Quantidade de notas de {i}: {quantidade_de_cedulas[i]}")
print(f"Total: {menor_quantidade_de_cedulas}")
```

12. Leia um inteiro positivo N, mostre todos os seus divisores positivos e informe quantos divisores ele possui.


```python
N = 0
while N <= 0:
  try:
    N = int(input("Digite um número inteiro positivo: "))
    break
  except ValueError:
    print("Digite um número válido.")

divisores_positivos = []
numero_divisores_positivos = 0
for i in range(1, N):
  if N % i == 0:
    divisores_positivos.append(i)
    numero_divisores_positivos += 1
print(f"Divisores positivos: {divisores_positivos}")
print(f"Número de divisores: {numero_divisores_positivos}")
```

13. Leia A e B positivos, com A < B, mostre os primos entre A e B (inclusive) e informe quantos foram encontrados.


```python
C = True
while C:
  try:
    A = int(input("Digite um número inteiro positivo: "))
    B = int(input("Digite um número inteiro positivo maior que o anterior: "))
    if A < B and A > 0 and B > 0:
      C = False
    else:
      print("Por favor, digite valores válidos.")
  except ValueError:
    print("Digite um valor válido.")
quantidade_de_primos = 0
for i in range(A, B + 1):
  primos = True
  for j in range(2, int((i ** 0.5)) + 1):
    if i % j == 0:
      primos = False
  if primos:
    quantidade_de_primos += 1
print(f"Quantidade de primos: {quantidade_de_primos}")
```

14. Considere o número secreto 37. O usuário terá no máximo 5 tentativas. Após cada erro, informe se o secreto é maior
ou menor que o palpite. Termine ao acertar ou ao usar as 5 tentativas.

```python
num_secreto = 37
quantidade_de_tentativas = 0
while quantidade_de_tentativas <= 5:
  while True:
    try:
      num = int(input("Digite a sua tentativa: "))
      break
    except ValueError:
     print("Digite um valor válido.")
  if num == num_secreto:
    print("Parabéns! Você acertou!")
    break
  elif num < num_secreto:
    print("O número secreto é maior.")
  else:
    print("O número secreto é menor.")
  quantidade_de_tentativas += 1
if quantidade_de_tentativas > 5:
  print("Fim do jogo! O número secreto era 37!")
```

15. Um caixa inicia o dia com R$ 200,00. Leia operações até que o código 0 seja informado. Código 1 = entrada de dinheiro; código 2 = retirada. Para cada operação válida, leia o valor. Uma retirada só é permitida se houver saldo suficiente. Ao final, informe: saldo final, quantidade de entradas, quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total que saiu.

```python
saldo = 200
quantidade_de_entradas = 0
quantidade_de_retiradas_realizadas = 0
quantidade_de_retiradas_recusadas = 0
codigo = "0"
while codigo != 0:
  while True:
    try:
      codigo = int(input("Digite o código 1 para entrada de dinheiro; 2 para retirada; 0 para encerrar: "))
      break
    except ValueError:
      print("Digite valor válido.")
  if codigo == 1:
    try:
      entrada = int(input("Digite o valor de entrada: "))
    except ValueError:
      print("Entrada inválida.")
    if entrada > 0:
      saldo += entrada
      quantidade_de_entradas += 1
    else:
      print("Entrada inválida.")
  elif codigo == 2:
    try:
      retirada = int(input("Digite o valor da retirada: "))
    except ValueError:
      print("Saída inválida.")
    if retirada > saldo:
      print("Saldo insuficiente.")
      quantidade_de_retiradas_recusadas += 1
    elif retirada < 0:
      print("Saída inválida.")
    else:
      saldo -= retirada
      quantidade_de_retiradas_realizadas += 1
  elif codigo != 0:
    print("Código inválido.")
  print("-----")
print(f"Saldo final: {saldo}")
print(f"Quantidade de entradas: {quantidade_de_entradas}")
print(f"Quantidade de retiradas realizadas: {quantidade_de_retiradas_realizadas}")
print(f'Quantidade de retiradas recusadas: {quantidade_de_retiradas_recusadas}')
```
