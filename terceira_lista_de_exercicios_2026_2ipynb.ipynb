1. Guardando e analisando notas

Leia 5 notas, armazene todas em uma lista e, somente depois das leituras, mostre as notas, a média e quantas notas são maiores ou iguais à média.

```python
notas = []
for i in range(5):
 nota = float(input("Digite a nota: "))
 notas.append(nota)
soma = 0
for nota in notas:
 soma += nota
media = soma / len(notas)
acima_ou_igual = 0
for nota in notas:
 if nota >= media:
  acima_ou_igual += 1
print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

4. Lista de números: pares e ímpares

6. Nomes sem repetição

Leia nomes até que o usuário digite FIM. Armazene somente nomes que ainda não estejam na lista. Ao final, mostre a lista e a quantidade de nomes diferentes cadastrados.

```python
nomes = []
FIM = False
while FIM == False:
  repeticao = False
  name = input("Digite o nome: ")
  if name == "FIM":
    break
  for nome in nomes:
    if name == nome:
      repeticao = True
  if repeticao == False:
    nomes.append(name)
print(f"Nomes: {nomes}")
print(f"Quantidade de nomes diferentes cadastrados: {len(nomes)}")
```

7. Dias da semana em uma tupla

Crie uma tupla contendo os sete dias da semana, em ordem. Leia um número de 1 a 7 e mostre o dia correspondente. Caso o número esteja fora desse intervalo, informe que a opção é inválida.

```python
dias_semana = ("domingo", "segunda-feira", "terça-feira", "quarta-feira", "quinta-feira", "sexta-feira", "sábado")
numero = int(input("Digite um número de 1 a 7: "))
if numero < 1 or numero > 7:
  print("Opção inválida.")
else:
  dia = dias_semana[numero - 1]
print(f"O dia correspondente ao número {numero} é: {dia}.")
```

8. Estatísticas de uma tupla

9. Dicionário de aluno

Cadastre nome, turma, nota1 e nota2 de um aluno em um dicionário. Calcule a média, acrescente ao próprio dicionário as chaves media e situacao e mostre todos os dados. Use: aprovado >= 7; recuperação >= 5; reprovado < 5.

```python
nome_do_aluno = input("Digite o nome do aluno: ")
turma_do_aluno = input("Digite a turma do aluno: ")
nota1 = float(input("Digite a primeira nota do aluno: "))
nota2 = float(input("Digite a segunda nota do aluno: "))
aluno = {"nome": nome_do_aluno, "turma": turma_do_aluno, "nota 1": nota1, "nota 2": nota2}
media = (nota1 + nota2) / 2
if media >= 7:
  situacao = "aprovado"
elif media >= 5:
  situacao = "recuperacao"
else:
  situacao = "reprovado"
media_situcao = {"média": media, "situação": situacao}
aluno.update(media_situcao)

print(f"Aluno: {aluno['nome']}.")
print(f"Turma: {aluno['turma']}.")
print(f"Nota 1: {aluno['nota 1']}.")
print(f"Nota 2: {aluno['nota 2']}.")
print(f"Média do aluno: {aluno['média']}.")
print(f"Situação do aluno: {aluno['situação']}.")
```

12. Estoque e atualização

Crie um dicionário com 4 produtos e suas quantidades em estoque. Leia o nome de um produto e uma quantidade vendida. Se o produto existir e houver estoque suficiente, atualize o dicionário; caso contrário, informe o problema. Mostre o estoque final.

```python
estoque = {"Melão": 35, "Uva": 120, "Melancia": 55, "Morango": 65}
produto_pesquisado = input("Digite o nome do produto: ")
q_vendida = int(input("Digite a quantidade vendida: "))
if produto_pesquisado in estoque:
  if q_vendida <= estoque[produto_pesquisado]:
    estoque[produto_pesquisado] -= q_vendida
  else:
    print("Quantidade vendida maior que disponível em estoque.")
else:
  print("Produto não está em estoque.")

print(f"Estoque final: {estoque}")
```

13. Cadastro de vários alunos

Cadastre 5 alunos. Para cada aluno, leia nome e duas notas, calcule a média e armazene nome, notas e média em um dicionário. Cada dicionário deve ser adicionado a uma lista. Ao final, mostre nome e média de todos os alunos.

```python
lista_alunos = []
for i in range(5):
  nome_do_aluno = input("Digite o nome do aluno: ")
  nota1 = float(input("Digite a primeira nota do aluno: "))
  nota2 = float(input("Digite a segunda nota do aluno: "))
  print("----------")
  media = round((nota1 + nota2) / 2, 2)
  dic_alunos = {"nome": nome_do_aluno, "nota 1": nota1, "nota 2": nota2, "média": media}
  lista_alunos.append(dic_alunos)
print("\n----------RELATÓRIO----------")
for item in lista_alunos:
  print(f"Nome: {item['nome']}")
  print(f"Média: {item['média']}")
  print("----------")
```

15. Mini sistema de estoque

Construa um pequeno cadastro de 5 itens de estoque. Cada item deve ser um dicionário com codigo, nome, quantidade e preco. Armazene os dicionários em uma lista. Depois: (a) mostre todos os itens; (b) calcule o valor total do estoque; (c) identifique o item de maior valor em estoque (quantidade × preço); (d) crie uma tupla resumo contendo (quantidade_de_itens_cadastrados, valor_total_estoque).

```python
lista_itens = []
valores_por_item = []
quantidade_de_itens_cadastrados = 0
for i in range(5):
  codigo = int(input("Digite o código do produto: "))
  nome = input("Digite o nome do produto: ")
  quantidade = int(input("Digite a quantidade do produto: "))
  print("----------")
  quantidade_de_itens_cadastrados += 1
  item = {"codigo": codigo, "nome": nome, "quantidade": quantidade, "preço": preco}
  lista_itens.append(item)

for item in lista_itens:
  valor = item['quantidade'] * item['preço']
  valores_por_item.append(valor)

for valor in valores_por_item:
  valor_total_estoque += valor

maior_valor = valores_por_item[0]
for i in range(5):
  if maior_valor < valores_por_item[i]:
    maior_valor = valores_por_item[i]
for item in lista_itens:
  if maior_valor in item:
    item_de_maior_valor = item['nome']

print(f"Todos os itens: {lista_itens} \n----------")
print(f"Valor total do estoque: {valor_total_estoque} \n----------")
print(f"Item de maior valor em estoque: {item_de_maior_valor}")
tupla_resumo = (quantidade_de_itens_cadastrados, valor_total_estoque)
```

Explicando o porquê de usar cada uma das três estruturas:


Usamos o dicionário para cada item pois o dicionário propicia uma organização por chaves, o que é muito útil para formar um índice de elementos.


Usamos a lista no código para armazenar os itens pois a lista é facilmente uma das práticas para o caso, pois, ao contrário das tuplas, é mutável; e, ao contrário dos dicionário, não precisa de chave. Também a usei pelos mesmos motivos para armazenar os valores por item e depois encontrar o valor total em estoque, o maior valor e o seu item.


Usamos a tupla para resumo pois ela é imutável, e, portanto, segura para armazenar informações que não precisam e não podem ser mudadas.
